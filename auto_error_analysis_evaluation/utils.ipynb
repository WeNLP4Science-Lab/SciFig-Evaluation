```python
import json

def extract_error_detail(json_file_path, field:str):
    components = set()

    # Load JSON data
    with open(json_file_path, 'r', encoding='utf-8') as f:
        data = json.load(f)

    # Handle both list of entries or single dict
    entries = data if isinstance(data, list) else [data]

    # Loop through each entry
    for entry in entries:
        error_detail = entry.get("error_detail", {})
        component = error_detail.get(field)
        if component:
            components.add(component)

    return components
```

```python

# Example usage
if __name__ == "__main__":
    file_path = "flattened_detected_errors.json"  # path to your JSON file
    component_result = extract_error_detail(file_path, "component")
    target_result = extract_error_detail(file_path, "target")
    print(component_result)
    print(target_result)
```
